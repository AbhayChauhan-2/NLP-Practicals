```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample text documents
document1 = "The quick brown fox jumps over the lazy dog."
document2 = "The quick brown cat leaps over the sleepy dog."
document3 = "A completely different sentence with no relation."

# Create a TfidfVectorizer object
# TfidfVectorizer converts a collection of raw documents to a matrix of TF-IDF features.
vectorizer = TfidfVectorizer()

# Fit the vectorizer to the documents and transform them into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform([document1, document2, document3])

# Calculate cosine similarity between document1 and document2
# The cosine_similarity function takes two vectors/matrices and returns their cosine similarity.
cosine_sim_doc1_doc2 = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
print(f"Cosine similarity between Document 1 and Document 2: {cosine_sim_doc1_doc2[0][0]:.4f}")

# Calculate cosine similarity between document1 and document3
cosine_sim_doc1_doc3 = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[2:3])
print(f"Cosine similarity between Document 1 and Document 3: {cosine_sim_doc1_doc3[0][0]:.4f}")

# You can also compute all pairwise similarities at once
pairwise_cosine_sim = cosine_similarity(tfidf_matrix)
print("\nPairwise Cosine Similarities:")
print(np.round(pairwise_cosine_sim, 4))

```
