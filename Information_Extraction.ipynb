```python

import spacy

nlp = spacy.load("en_core_web_sm")

text = """
Aditi Panwar is studying Computer Science at NIET in Greater Noida.
She is interested in Artificial Intelligence and works with Microsoft
and Google technologies.
"""

doc = nlp(text)

print("Named Entities:\n")

for ent in doc.ents:
    print(ent.text, "---->", ent.label_)
for ent in doc.ents:
  print(f"Entity: {ent.text}")
  print(f"Type: {ent.label_}")
  print(f"Explanation: {spacy.explain(ent.label_)}")
  print()
```
